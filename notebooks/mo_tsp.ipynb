```python
import random

import matplotlib.pyplot as plt
import numpy as np

from genetic.algorithms import GeneticAlgorithmMO
from genetic.configurations import ProblemConfig
from genetic.crossover import OrderCrossover
from genetic.initialization import PermutationInitialization
from genetic.mutation import SwapMutation
from genetic.problems import MOTSProblem
from genetic.replacement import GenerationalReplacement
from genetic.selection import ParetoSelection
```

```python
plt.style.use("dark_background")
random.seed(42)
np.random.seed(42)


def create_figure(w=12, h=8):
    fig = plt.figure(figsize=(w, h))
    ax = fig.add_subplot(111, projection="3d")
    return fig, ax


def plot_nodes(ax, nodes, elevations, color="r", size=5):
    x_coords, y_coords = zip(*nodes)
    ax.scatter(x_coords, y_coords, elevations, c=color, s=size)


def plot_path(ax, path, elevations, color="b", linewidth=1.5):
    x_path, y_path = zip(*path)
    ax.plot(x_path, y_path, elevations, color=color, linewidth=linewidth)
```

```python
config = ProblemConfig(max_evaluations=100000, seed=42, output_dir="results/tsp")

# Problem
problem = MOTSProblem(config, cities, perlin_seed=42)

# Operators
initialization = PermutationInitialization()
selection = ParetoSelection()
crossover = OrderCrossover()
mutation = SwapMutation(mutation_rate=0.15)
replacement = GenerationalReplacement()

# Genetic Algorithm
ga = GeneticAlgorithmMO(
    problem=problem,
    population_size=50,
    initialization=initialization,
    selection=selection,
    crossover=crossover,
    mutation=mutation,
    replacement=replacement,
)
```

```python
from mpl_toolkits.mplot3d import Axes3D

x_coords, y_coords = zip(*cities)
z_coords = problem.elevations

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")
scatter = ax.scatter(x_coords, y_coords, z_coords, c=z_coords, cmap="terrain", s=50)

ax.set_xlabel("Coordenada X")
ax.set_ylabel("Coordenada Y")
ax.set_zlabel("Elevación")
ax.set_title("Representación 3D de Ciudades")
plt.colorbar(scatter, label="Altura")
```

```python
from matplotlib.pyplot import imshow
from matplotlib.pyplot import subplots

fig, axes = subplots(1, 2)
axes[0].imshow(problem.dist_matrix)
axes[1].imshow(problem.time_matrix)
```

```python
# Al final del todo calcula el frente de Pareto final y suele tardar bastante tiempo (16'5 min aprox)
result = ga.run()
```
