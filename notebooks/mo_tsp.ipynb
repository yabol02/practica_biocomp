# Base result NSGA-II

```python
from pymoo.core.problem import Problem
import random
import numpy as np
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.optimize import minimize

class TSPProblem(Problem):
    def __init__(self, n_cities, seed=50):
        # Define the number of variables (cities) and objectives (time and distance)
        super().__init__(n_var=n_cities, n_obj=2, n_constr=0, xl=0, xu=1)

        # Initialize the random seed for reproducibility
        if seed is not None:
            np.random.seed(seed)

        # Generate random distances and times between cities
        self.distances = np.random.uniform(0, 100, size=(n_cities, n_cities))
        self.distances = (self.distances + self.distances.T) / 2
        self.times = np.random.uniform(0, 300, size=(n_cities, n_cities))
        random.seed(None)

        # Ensure the diagonal is zero (distance/time from a city to itself is zero)
        np.fill_diagonal(self.distances, 0)
        np.fill_diagonal(self.times, 0)

    def _evaluate(self, x, out, *args, **kwargs):
        # Calculate the total distance and time for each solution in x
        distances = []
        times = []

        for route in x:
            perm = np.argsort(route)
            total_distance = 0
            total_time = 0

            for i in range(len(perm) - 1):
                total_distance += self.distances[perm[i], perm[i + 1]]
                total_time += self.times[perm[i], perm[i + 1]]

            # Add distance and time from last city back to the first city
            total_distance += self.distances[perm[-1], perm[0]]
            total_time += self.times[perm[-1], perm[0]]

            distances.append(total_distance)
            times.append(total_time)

        out["F"] = np.column_stack([distances, times])

problem = TSPProblem(100, seed=42)
print(np.shape(problem.distances))
print(np.shape(problem.times))
algorithm = NSGA2(pop_size=100)
algorithm2 = NSGA2(pop_size=50)
res_batir_1 = minimize(problem, algorithm, ("n_gen", 100), verbose=False)
res_batir_2 = minimize(problem, algorithm2, ("n_gen", 200), verbose=False)
```

```python
import random

import matplotlib.pyplot as plt
import numpy as np

from genetic.algorithms import GeneticAlgorithmMO
from genetic.configurations import ProblemConfig
from genetic.crossover import OrderCrossover
from genetic.initialization import PermutationInitialization
from genetic.mutation import SwapMutation
from genetic.problems import MOTSProblem
from genetic.replacement import GenerationalReplacement
from genetic.selection import ParetoSelection
```

```python
plt.style.use("dark_background")
random.seed(42)
np.random.seed(42)


def create_figure(w=12, h=8):
    fig = plt.figure(figsize=(w, h))
    ax = fig.add_subplot(111, projection="3d")
    return fig, ax


def plot_nodes(ax, nodes, elevations, color="r", size=5):
    x_coords, y_coords = zip(*nodes)
    ax.scatter(x_coords, y_coords, elevations, c=color, s=size)


def plot_path(ax, path, elevations, color="b", linewidth=1.5):
    x_path, y_path = zip(*path)
    ax.plot(x_path, y_path, elevations, color=color, linewidth=linewidth)
```

## Try 4: Configuración Optimizada
Combinación de operadores óptimos para TSP:
- **EdgeRecombinationCrossover (ERX)**: Preserva el máximo número de aristas de ambos padres
- **InversionMutation**: Equivalente a 2-opt, elimina cruces y mejora estructura geométrica
- **NeighborInitialization**: Inicialización heurística con k-nearest neighbors para mejor calidad inicial
- **Mayor población**: 150 individuos para mejor exploración del frente de Pareto

```python
from genetic.crossover import EdgeRecombinationCrossover, PMXCrossover
from genetic.initialization import NeighborInitialization, DiverseNNInitialization
```

```python
config = ProblemConfig(max_evaluations=100000, seed=42, output_dir="results/tspmo_optimized")

problem = MOTSProblem(config, cities)

# Operadores optimizados para TSP multiobjetivo
initialization = NeighborInitialization(k_best=3)  # Inicialización heurística con diversidad
selection = ParetoSelection()
crossover = EdgeRecombinationCrossover()  # Preserva aristas - ideal para TSP
mutation = InversionMutation(mutation_rate=0.2)  # 2-opt implícito - muy efectivo
replacement = GenerationalReplacement()

# Genetic Algorithm con población más grande
ga = GeneticAlgorithmMO(
    problem=problem,
    population_size=150,
    initialization=initialization,
    selection=selection,
    crossover=crossover,
    mutation=mutation,
    replacement=replacement,
)

result = ga.run()
scatter = compare_with_base(result, res_batir_1, res_batir_2)
scatter.show()
```

## Try 5: PMX + Mutación Combinada
Alternativa con PMXCrossover que también es muy efectivo para TSP, combinando dos mutaciones.

```python
from genetic.mutation import CombinedMutation

config = ProblemConfig(max_evaluations=28000, seed=42, output_dir="results/tspmo_pmx")

problem = MOTSProblem(config, cities)

# Operadores con PMX y mutación combinada
initialization = NeighborInitialization(k_best=5)  # Mayor diversidad en inicialización
selection = ParetoSelection()
crossover = PMXCrossover()  # Partially Mapped Crossover - muy efectivo para TSP

# Mutación combinada: primero inversión (explotación), luego swap (exploración)
mutation = CombinedMutation([
    InversionMutation(mutation_rate=0.15),  # 2-opt para mejora local
    SwapMutation(mutation_rate=0.05),        # Diversificación ocasional
])
replacement = GenerationalReplacement()

# Genetic Algorithm
ga = GeneticAlgorithmMO(
    problem=problem,
    population_size=120,
    initialization=initialization,
    selection=selection,
    crossover=crossover,
    mutation=mutation,
    replacement=replacement,
)

result = ga.run()
scatter = compare_with_base(result, res_batir_1, res_batir_2)
scatter.show()
```

## Try 6: Configuración Óptima Final
Combinación final optimizada para TSP multiobjetivo:
- **DiverseNNInitialization**: Inicialización determinista con diversidad máxima (cada individuo empieza en ciudad diferente)
- **EdgeRecombinationCrossover**: Preserva aristas de ambos padres - el mejor para TSP
- **CombinedMutation**: InversionMutation (2-opt) + SwapMutation para balance explotación/exploración
- **Población grande (200)**: Mayor diversidad en el frente de Pareto

```python
config = ProblemConfig(max_evaluations=100000, seed=42, output_dir="results/tspmo_final")

problem = MOTSProblem(config, cities)

# Operadores óptimos para TSP multiobjetivo
initialization = DiverseNNInitialization()  # Máxima diversidad - cada individuo inicia en ciudad diferente
selection = ParetoSelection()
crossover = EdgeRecombinationCrossover()  # Preserva aristas - el mejor crossover para TSP

# Mutación combinada optimizada
mutation = CombinedMutation([
    InversionMutation(mutation_rate=0.25),  # 2-opt agresivo para explotación
    SwapMutation(mutation_rate=0.08),        # Diversificación moderada
])
replacement = GenerationalReplacement()

# Genetic Algorithm con población grande para mejor cobertura del frente Pareto
ga = GeneticAlgorithmMO(
    problem=problem,
    population_size=200,
    initialization=initialization,
    selection=selection,
    crossover=crossover,
    mutation=mutation,
    replacement=replacement,
)

result = ga.run()
print_result(result)
scatter = compare_with_base(result, res_batir_1, res_batir_2)
scatter.show()
```
