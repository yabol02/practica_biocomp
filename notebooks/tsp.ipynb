```python
import random

import matplotlib.pyplot as plt
import numpy as np

from genetic.algorithms import GeneticAlgorithmSO
from genetic.configurations import ProblemConfig
from genetic.crossover import OrderCrossover
from genetic.initialization import PermutationInitialization
from genetic.mutation import SwapMutation
from genetic.problems import TSProblem
from genetic.replacement import ElitistReplacement
from genetic.selection import TournamentSelection
```

```python
plt.style.use("dark_background")
random.seed(42)
np.random.seed(42)


def create_figure(w=12, h=8):
    fig, ax = plt.subplots(figsize=(w, h))
    ax.axis("off")
    return fig, ax


def plot_nodes(ax, nodes, color="r", size=5):
    for x, y in nodes:
        ax.plot(x, y, ".", color=color, markersize=size)


def plot_path(ax, path, color="b", linewidth=1.5):
    for i in range(len(path) - 1):
        x = (path[i][0], path[i + 1][0])
        y = (path[i][1], path[i + 1][1])
        ax.plot(x, y, color=color, linewidth=linewidth)
```

```python
config = ProblemConfig(max_evaluations=1000000, seed=38, output_dir="results/tsp")

# Problem
problem = TSProblem(config, cities)

# Operators
initialization = PermutationInitialization()
selection = TournamentSelection(tournament_size=50)
crossover = OrderCrossover()
mutation = SwapMutation(mutation_rate=1)
replacement = ElitistReplacement(elite_size=15)

# Genetic Algorithm
ga = GeneticAlgorithmSO(
    problem=problem,
    population_size=100,
    initialization=initialization,
    selection=selection,
    crossover=crossover,
    mutation=mutation,
    replacement=replacement,
    print_interval=1000,
)
```

```python
print("=" * 80)
print("OPTIMIZING TRAVELING SALESMAN PROBLEM")
print("=" * 80)
result = ga.run()
```

```python
print("\n" + "=" * 40)
print("RESULTS")
print("=" * 40)
print(f"Problem: {result.problem_name}")
print(f"Best fitness: {result.best_fitness:.8f}")
print(f"Best solution: cities={result.best_solution}")
print(f"Best distance: {problem._sol_distance(result.best_solution):.8f}")
print(f"Evaluations used: {result.evaluations_used}")
print(f"Generations: {len(result.history)}")
```
