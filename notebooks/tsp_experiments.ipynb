## Plots

```python
plt.style.use("dark_background")
random.seed(42)
np.random.seed(42)


def create_figure(w=12, h=8):
    fig, ax = plt.subplots(figsize=(w, h))
    ax.axis("off")
    return fig, ax


def plot_nodes(ax, nodes, color="r", size=5):
    for x, y in nodes:
        ax.plot(x, y, ".", color=color, markersize=size)


def plot_path(ax, path, color="b", linewidth=1.5):
    for i in range(len(path) - 1):
        x = (path[i][0], path[i + 1][0])
        y = (path[i][1], path[i + 1][1])
        ax.plot(x, y, color=color, linewidth=linewidth)
```

```python
# Plot evolution
def plot_evolution(mean_history, best_history, worst_history):
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(best_history, label="Best", color="cyan")
    ax.plot(mean_history, label="Mean", color="yellow")
    ax.plot(worst_history, label="Worst", color="red")
    ax.set_xlabel("Generation")
    ax.set_ylabel("Fitness")
    ax.set_title("Evolution of Fitness over Generations")
    ax.set_ylim(bottom=0)
    ax.legend()
    plt.show()

def plot_best(best_solution, best_fitness):
    cities_solution = [cities[idx] for idx in best_solution]
    cities_solution.append(cities_solution[0])

    fig, ax = create_figure()
    plot_nodes(ax, cities)
    plot_path(ax, cities_solution)
    fig.text(0.5, 0.02, f"Total distance: {best_fitness:.4f}", ha="center")
```

```python
from genetic.optuna import OptunaOptimizer
from optuna.samplers import TPESampler
config = ProblemConfig(max_evaluations=1000000, seed=38, output_dir="results/tsp")

# Problem
problem = TSProblem(config, cities, minimize=True)

# Optuna
optuna_trials = 100
startup_trials = int(0.1 * optuna_trials)
sampler = TPESampler(
    n_startup_trials=startup_trials,
    multivariate=True,
    seed=42,
)

optimizer = OptunaOptimizer(problem_config=config,
                            problem=problem,
                            total_trials=optuna_trials,
                            sampler=sampler,
                            max_generations=100,
                            conf_repetitions=2
                            )
```
