```python
from classes.algorithms import GeneticAlgorithmSO
from classes.configurations import ProblemConfig
from classes.crossover import BlendCrossover
from classes.initialization import RandomInitialization
from classes.mutation import UniformMutation
from classes.problems import HimmelblauProblem
from classes.selection import TournamentSelection
```

```python
config = ProblemConfig(
    max_evaluations=3500,
    seed=42,
    output_dir="results/himmelblau"
)

# Problem
problem = HimmelblauProblem(config)

# Operators
initialization = RandomInitialization()
selection = TournamentSelection(tournament_size=3)
crossover = BlendCrossover(alpha=0.5)
mutation = UniformMutation(mutation_rate=0.15, bounds=problem.get_bounds())
# TODO: Add a replace/suvival operator (?)

# Genetic Algorithm
ga = GeneticAlgorithmSO(
    problem=problem,
    population_size=50,
    initialization=initialization,
    selection=selection,
    crossover=crossover,
    mutation=mutation
)
```

```python
print("=" * 80)
print("OPTIMIZING HIMMELBLAU FUNCTION")
print("=" * 80)
result = ga.run()
```

```python
print("\n" + "=" * 80)
print("RESULTS")
print("=" * 80)
print(f"Problem: {result.problem_name}")
print(f"Best fitness: {result.best_fitness:.8f}")
print(f"Best solution: x={result.best_solution[0]:.6f}, y={result.best_solution[1]:.6f}")
print(f"Evaluations used: {result.evaluations_used}")
print(f"Generations: {len(result.history)}")
print(f"\nKnown global minima (f=0):\n  (3.0, 2.0)\n  (-2.805118, 3.131312)\n  (-3.779310, -3.283186)\n  (3.584428, -1.848126)")
```
